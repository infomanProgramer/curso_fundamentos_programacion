# Descomposición de números

### Ejemplos de descomposición

```python
# Descomposición de derecha a izquierda
import math

n = 37482

while n != 0:
    d = n % 10
    print(d)
    n = math.trunc(n / 10)
    print(n)
```

```python
# Descomposción de izquierda a derecha
n = 37482

while n != 0:
    num_digitos = math.floor(math.log10(n)) + 1
    d = math.trunc(n/(10 ** (num_digitos - 1)))
    print(d)
    n = n % 10 ** (num_digitos - 1)
    #print(n)
```

## Composición de números

```python
# Composición de izquierda a derecha
n = 37482
new = 0
while n != 0:
    num_digitos = math.floor(math.log10(n)) + 1
    d = math.trunc(n/(10 ** (num_digitos - 1)))
    print(d)
    n = n % 10 ** (num_digitos - 1)
    #Composición
    new = new * 10 + d
    #print(n)
print(new)
```

```python
# Composición de derecha a izquierda
n = 37482
new = 0
posicion = 1
while n != 0:
    d = n % 10
    #print(d)
    n = math.trunc(n / 10)
    #Composición
    new = d * posicion + new
    posicion = posicion * 10
print(new)
```
