# Estructuras de control

Las estructuras de control son simplemente una forma de especificar el flujo de control en los programas. Básicamente, analizan y eligen en qué dirección fluye un programa en función de ciertos parámetros o condiciones. Existen tres tipos básicos de lógica o flujo de control, conocidos como:

1. Lógica de secuencia o flujo secuencial
2. Lógica de selección o flujo condicional
3. Lógica iterativa o flujo repetitivo

## Flujo secuencial

La lógica secuencial, como su nombre indica, sigue un flujo serial o secuencial en el que el flujo depende de la serie de instrucciones dadas a la computadora. A menos que se den nuevas instrucciones, los módulos se ejecutan en la secuencia obvia. Las secuencias pueden especificarse explícitamente mediante pasos numerados. También se sigue implícitamente el orden en que se escriben los módulos. La mayor parte del procesamiento, incluso algunos problemas complejos, generalmente seguirán este patrón de flujo elemental.

## Lógica de selección o flujo condicional

La lógica de selección simplemente implica una serie de condiciones o parámetros que deciden uno de entre varios módulos escritos. Las estructuras que utilizan este tipo de lógica se conocen como estructuras condicionales. Estas estructuras pueden ser de tres tipos:

- **De una sola alternativa**

~~~
if condicion then
  [Modulo A]
~~~

image.png

- **De doble alternativa**

La estructura tiene la siguiente forma:

~~~
If (Condition), then:
     [Module A]
Else:
     [Module B]
[End if structure]
~~~

image.png

- Multiples alternativas

~~~
If (condition A), then:
     [Module A]
Else if (condition B), then:
     [Module B]
Else if (condition C), then:
     [Module C]
[End If structure]
~~~

image.png



## Logica de iteración o flujo repetitivo

La lógica de iteración emplea un bucle que consta de una instrucción de repetición seguida de un módulo conocido como el cuerpo del bucle. Los dos tipos de estas estructuras son:

- **Estrucutura de repetición FOR**

~~~
FOR i = 1; i <= 10; i = i + 1
  MOSTRAR i
END FOR
~~~

image.png

Resultado
~~~

~~~

- **Estructura de repetición WHILE**

~~~
i = 1
REPEAT WHILE i < = 10 THEN
     MOSTRAR i
     i = i + 1
END WHILE
~~~

## Variables

En programación, a menudo necesitamos una ubicación de almacenamiento con nombre para guardar datos o valores. Mediante variables, podemos almacenar estos datos en nuestro programa y acceder a ellos posteriormente.

### ¿Qué es una variable en programación?

En programación, una variable es una ubicación de almacenamiento con nombre que guarda un valor o dato. Estos valores pueden cambiar durante la ejecución de un programa, de ahí el término "variable". Las variables son esenciales para almacenar y manipular datos en programas informáticos. Una variable es el componente básico de un programa que se puede usar en expresiones como sustituto del valor que almacena.

### Declaraciones de variables

En programación, la declaración de variables implica especificar el tipo y el nombre de una variable antes de usarla en el programa. La sintaxis puede variar ligeramente entre lenguajes de programación, pero el concepto fundamental se mantiene.

## Acumuladores y Contadores

- **Contador:** cuenta cuantas veces ocurre algo
- **Acumulador:** Suma o acumula valores

#### Contador

Es una variable que aumenta o disminuye generalemente de 1 en 1 para llevar la cuenta de algo

```python
contador = 0
i = 1
while i <= 5:
  print(i)
  i = i + 1
```

#### Ejercicio

Contar los números pares

```python
contador = 0
i = 1
while i <= 5:
  if i % 2 == 0:
    contador = contador + 1
  i = i + 1
print(f"La cantidad de números pares en el intervalo de 1 a 5 es de {contador}")
```

#### Acumulador

Un acumulador es una variable que va guardando una suma progresiva

#### Ejemplo

Sumar cinco números

```python
suma = 0
i = 1
while i <= 5:
  suma = suma + i
  i = i + 1
print(f"La suma es {suma}")
```
