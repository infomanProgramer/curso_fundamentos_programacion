# Introducción

Vamos aprender a resolver problemas mediante algoritmos, se aprendera sintaxis

> *Problema -> Analisis -> Algoritmo -> Pseudocódigo -> Diagrama de flujo -> Código*

## ¿Qué es programación?

La programación tambien conocida como codificación, es el proceso de crear un conjunto de instrucciones para decirle a una computadora como ejecutar una tarea especifica. Estas instrucciones llamadas programas estan escritas en un lenguaje que la computadora pueda ejecutar y  entender.

Pensemos que un programa son comandos que se le da a un robot. Nostros le decimos al robot que hacer paso a paso como una instrucción.

El proposito de la programación es resolver problemas y automatizar tareas. Existen una gran cantidad de programas y aplicaciones, desde progrmas que relizan tareas simples hasta prograamas que realizan tareas complejas como los videojuegos o un gestor de base de datos.

### ¿Como funciona la programación?

La programación involucra varios pasos:

- **Definir lo que va hacer el programa:** Definir el problema que queremos resolver y que queremos que logre el problema
- **Diseño de algoritmos:** Desarrollar paso a paso el procedimiento para resolver el problema
  - Pseudocódigo
  - Diagrama de flujo
- **Codificar:** Traducir el algoritmo a un lenguaje de programación usando un editor de código o un entorno de desarrollo integrado (IDE)
- **Prueba y depuración:** Ejecutar el programa e identificar y arreglar errores
- **Despliegue:** Publicar o implementar el programa en un entorno donde pueda ser usado por los usuarios

## Beneficios de aprender a programar

Aprender a programar ofrece multiples beneficios tanto en lo personal como en los profesional

- **Desarrollo del pensamiento critico y habilidades para resolver problemas:** La programación fomenta el pensamiento lógico, la descomposición de problemas y la búsqueda de soluciones creativas.
- **Potencia tu creatividad e innovación:** La programación te permite crear tus propias herramientas y aplicaciones, convirtiendo tus ideas en realidad.
- **Incrementa tu empleabilidad:** La demanda de programadores cualificados es alta y está creciendo en diversos sectores.
- **Mejora tus habilidades de comunicación y colaboración:** Trabajar con código requiere colaboración y comunicación clara
- **Ganar un profundo entendimiento en tecnología:** Aprender como funciona el código te da un mejor entendimiento de como las computadoras trabajan y de como ellas son usadas en tu entorno
- **Desarrolla la autoconfianza y la motivación:** Completar proyecto de programación con exito puede aumentar tu confianza y motivarte a aprender cosas nuevas

## ¿Qué clase de problemas resuelven los algoritmos?

- Todo el mundo todos quieren acceder y recuperar grandes cantidades de información. Por ejemplo un motor de búsqueda para encontrar rápidamente paginas que contengan información especifica
- El comercio eletrónico permite negociar e intercambiar bienes y servicios electronicamente y depende de la privacidad de la información como numero de tarjetas de crédito, contraseñas, extractos bancarios, etc. Las tecnologias principales utilizadas en el comercio electronico incluyen criptografía de clave pública y las firmas digitales, las cuales se basan en algoritmos númericos y teoria de números.
- Las empresas en general necesitan asignar recursos escasos de manera beneficiosa como:
    - Una compañia petrolera podria querer saber donde ubicar sus pozos para maximizar ganancias esperadas
    - Un candidato político podria querer saber donde invertir en publicidad para aumentar sus probabilidades de ganar las elecciones
    - Un proveedor de servicios de internet podria querer determinar donde ubicar recursos adicionales para atender a sus clientes de manera más eficaz.
    - Un banco podria querer determinar el mejor lugar para abrir una nueva sucursal.
- En un mapa de carreteras donde se marca la distancia entre cada par de intersecciones adyacentes deseas determinar la ruta mas corta de una intersección a otra. El número de rutas puede ser enorme
