# Contenido Fundamentos de Programación

## ¿A quien va dirigido este curso?

Este curso está dirigido a estudiantes de los primeros semestres de carreras como Informática, Ingeniería de Sistemas, Ingeniería de Software, Ingeniería Electrónica y Telecomunicaciones que deseen adquirir una base sólida en los fundamentos de la programación y el desarrollo de algoritmos.

Asimismo, está orientado a personas sin experiencia previa en programación y a profesionales de otras áreas que utilizan sistemas informáticos en su actividad laboral y desean comprender los principios de funcionamiento del software, fortalecer su pensamiento lógico y desarrollar las competencias necesarias para iniciarse en el desarrollo de aplicaciones.

## Prerrequisitos

- Manejo básico de computadoras
- Manejo básico de Windows a nivel usuario
- Conocimientos básicos de matemáticas
- No se requieren conocimientos previos de programación

## Herramientas

- Visual Studio Code
- Python
- Herramienta de inteligencia artificial generativa como ChatGPT, GitHub Copilot o Google Gemini

## Horario

Martes y jueves de 20:30 a 22:30

## Contenido

- Introducción a la programación y algoritmos (3 hrs)
    - Algoritmos
    - Pseudocódigo
    - Diagramas de flujo
- Fundamentos de programación (3 hrs)
    - Variables
    - Constantes
    - Tipos de datos
- Estructuras de control (5 hrs)
    - Condicionales
    - Ciclos
- Modularización (3 hrs)
    - Funciones
    - Procedimientos
    - Parámetros
- Arreglos y cadenas (3 hrs)
    - Arreglos unidimensionales
    - Recorrido
    - Búsqueda
- Patrones algorítmicos (3 hrs)
    - Contadores
    - Acumuladores
    - Validaciones
    - Maximos, minimos y promedios
- Introducción al análisis de algoritmos (2 hr)
    - Eficiencia
    - Complejidad
    - Notación Big O
- IA como ayoyo al aprendisaje de programación (2hrs)
    - IA generativa
    - Prompts
    - Explicación
    - Depuración y validacion de respuestas

## Ponderación

- Participación en clase 30%
- Asistencia 10%
- Proyecto final 60%
- Total 100%
