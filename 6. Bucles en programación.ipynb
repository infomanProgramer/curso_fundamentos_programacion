# Bucles en programación

**Bucles o instrucciones de iteración** en programación, son útiles cuando necesitamos repetir una tarea específica. Son esenciales, ya que reducen horas de trabajo a segundos.

## ¿Qué son los bucles en programación?

Los bucles, también conocidos como instrucciones iterativas, se utilizan cuando necesitamos ejecutar un bloque de código repetidamente. En programación, los bucles son estructuras de control de flujo que permiten la ejecución repetida de un conjunto de instrucciones o un bloque de código mientras se cumpla una condición específica.

## Tipos de bucles en programación

En programación, los bucles se clasifican en dos tipos principales según su mecanismo de control: bucles controlados por entrada y bucles controlados por salida.

1. **Bucles controlados por entrada:** En los bucles controlados por entrada, la condición de prueba se verifica antes de entrar en el cuerpo principal del bucle. Los bucles For y While son bucles controlados por entrada.

```python
# Bucle for controlador por entrada
for i in range(5):
    print(i, end=" ")
```

```python
# Bucle while controlado por entrada
i = 0
while(i < 5):
  print(i)
  i += 1
```

2. **Bucles controlados por salida:** En los bucles controlados por salida, la condición de prueba se evalúa al final del cuerpo del bucle. El cuerpo del bucle se ejecutará al menos una vez, independientemente de si la condición es verdadera o falsa. El bucle **do-while** es un ejemplo de bucle controlado por salida.

~~~java
int i = 0;
do {
    System.out.print(i + " ");
    i++; //i = i + 1
} while (i < 5);
~~~

```python
# Salida de un bucle do-while controlado
i = 0
while True:
  print(i, end=" ")
  i += 1 # i = i + 1
  if i >= 5:
    break
```

### Bucle For:

En programación, un bucle `for` es una estructura de control de flujo que itera sobre una secuencia de elementos, como un rango de números, elementos de una lista o caracteres de una cadena. El bucle está controlado por entrada, ya que determina el número de iteraciones antes de entrar en él.

### Bucle While:

En programación, un bucle while es una estructura de flujo de control con entrada controlada que ejecuta repetidamente un bloque de código mientras se cumpla una condición específica. El bucle continúa iterando mientras la condición sea verdadera y finaliza cuando la condición se evalúa como falsa.

### Bucle Do-While:

En programación, un bucle do-while es una estructura de flujo de control con salida que ejecuta un bloque de código al menos una vez y luego lo repite mientras una condición específica se mantenga verdadera. La característica distintiva de un bucle do-while es que la condición se evalúa después del bloque de código, lo que garantiza que este se ejecute al menos una vez, incluso si la condición es inicialmente falsa.

### Bucles anidados

En programación, los bucles anidados consisten en colocar un bucle dentro de otro. Esto permite la iteración de uno o más bucles dentro del cuerpo de otro bucle. Cada vez que se ejecuta el bucle externo, el bucle interno completa su iteración. Los bucles anidados se utilizan comúnmente para tareas que implican el procesamiento de datos multidimensionales, la impresión de patrones o el manejo de estructuras de datos complejas. El uso eficiente de bucles anidados es esencial, considerando el impacto potencial en la legibilidad del código y el tiempo de ejecución para conjuntos de datos grandes.

```python
for i in range(3):
  j = 0
  while j < 5:
    print(f"i = {i} j = {j} ", end="")
    j += 1
    print()
```
