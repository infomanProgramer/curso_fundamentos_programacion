# **10. Arreglos y cadenas**

Los arrays son una de las estructuras de datos fundamentales en programación, y Python ofrece varias maneras de trabajar con ellos. Comprender los arrays puede suponer un cambio radical para gestionar colecciones de datos de forma eficiente.

### **¿Qué son los arrays en Python?**

En Python, los arrays son colecciones ordenadas de elementos que pueden almacenar datos del mismo tipo. A diferencia de las listas (que son más flexibles), los arrays propiamente dichos en Python son más eficientes en cuanto al uso de memoria y más rápidos para las operaciones numéricas.

Python ofrece diferentes maneras de trabajar con matrices:

- Módulo de arreglos integrado de Python
- Arreglos NumPy (los más utilizados)
- Listas (el tipo de secuencia predeterminado de Python)

### **Método 1**: Uso del módulo Array integrado de Python

El módulo array proporciona un objeto array más compacto que las listas para almacenar tipos numéricos básicos.

```python
import array as arr

# Creación de una matriz de enteros
# 'i' representa un entero con signo

numbers = arr.array('i', [1, 2, 3, 4, 5])

print(numbers)  # Output: array('i', [1, 2, 3, 4, 5])
print(type(numbers))  # Output: <class 'array.array'>
```

Códigos de tipos de matrices comunes:

- 'b': carácter con signo (entero entre -128 y 127)
- 'i': entero
- 'f': float
- 'd': double (float con mayor precisión)

### **Operaciones básicas con el módulo de Array**

```python
import array as arr

# Creando un array de enteros
sales_figures = arr.array('i', [10500, 15200, 8700, 19300, 12600])

# Accediendo a sus elementos
print(f"Ventas de enero: {sales_figures[0]}")

# Actualizando valores
sales_figures[1] = 100
print(f"Actualizando ventas de febrero: {sales_figures}")

# Añadiendo elementos
sales_figures.append(14800)
sales_figures.append(1000)
print(f"Added June sales: ${sales_figures[-1]}")
print(sales_figures)

# Eliminando elementos
removed_value = sales_figures.pop(2)
print(f"Eliminando ventas de Marzo ({removed_value}) due to data error")
print(sales_figures)
```

### **Método 2**: Uso de arrays NumPy (recomendado)

Los arrays de NumPy son los arrays más utilizados en Python, especialmente en ciencia de datos y computación numérica.

```python
import numpy as np

temperatures = np.array([68, 71, 73, 69, 75, 83, 79])
print(temperatures)

monthly_sales = np.array([
    [5000, 6000, 7000],  # Q1 (Jan, Feb, Mar)
    [7500, 8200, 9000],  # Q2 (Apr, May, Jun)
    [9500, 9800, 10000], # Q3 (Jul, Aug, Sep)
    [12000, 15000, 13000] # Q4 (Oct, Nov, Dec)
])

print(monthly_sales)
```

### **Creación de matrices NumPy especiales**

```python
# Array de ceros
zero_array = np.zeros(5)
print(f"Zeros array: {zero_array}")

# Array de unos
ones_array = np.ones(5)
print(f"Ones array: {ones_array}")

# Rango de un array
range_array = np.arange(0, 10, 2)  # Inicio, Finaliza, Secuencia de pasos
print(f"Rango de un array: {range_array}")

# Disposición espaciada uniformemente
linspace_array = np.linspace(0, 1, 5)  # Inicio, Fin, número de elementos
print(f"Linspace array: {linspace_array}")  # Output: [0.   0.25 0.5  0.75 1.  ]
```

### **Operaciones con arrays NumPy**

```python
# Creando arrays de temperaturas de las ciudades de US (°F)

new_york = np.array([45, 48, 52, 58, 65])
los_angeles = np.array([68, 70, 72, 74, 76])

# Adición
temperature_sum = new_york + los_angeles
print(f"Sum of temperatures: {temperature_sum}")

# Multiplicación (elemento por elemento)
temperature_product = new_york * 2  # Double all temperatures
print(f"New York temperatures doubled: {temperature_product}")

# Operaciones estadisticas
print(f"Average NY temperature: {np.mean(new_york)}°F")
print(f"Maximum LA temperature: {np.max(los_angeles)}°F")
print(f"Minimum NY temperature: {np.min(new_york)}°F")
print(f"Standard deviation of NY temperatures: {np.std(new_york):.2f}°F")
```

### Remodelación y segmentación de arrays

```python
# Creando array de una dimensión
quarterly_sales = np.array([55000, 68000, 72000, 98000, 62000, 75000, 83000, 91000])

# Reorganizar a 2D (2 filas, 4 columnas) - 2 años de datos trimestrales
sales_by_year = quarterly_sales.reshape(2, 4)
print("Ventas por año (quarterly):")
print(sales_by_year)

# Dividiendo los arrays
print("\nVentas trimestrales del primer año:")
print(sales_by_year[0])  # Primera fila

print("\nQ2 ventas para ambos años:")
print(sales_by_year[:, 1])  # Segunda columna
```

### **Método 3:** Usar listas de Python como arreglos

Si bien no son arreglos propiamente dichos, las listas de Python pueden servir como estructuras similares a arreglos para muchos propósitos.

```python
# Creando una lista como un array
us_cities = ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"]

# Añadiendo elementos
us_cities.append("Philadelphia")

# Accediendo a los elementos
print(f"Tercera ciudad más grande: {us_cities[2]}")  # Output: Chicago

# Dividiendo
east_coast_cities = us_cities[0:2]
print(f"Ciudades de la costa Este: {east_coast_cities}")  # Output: ['New York', 'Los Angeles']
```

### Cuándo usar cada tipo de arreglo

- Utilice el módulo array de Python cuando:
  - Necesitas una colección simple de datos numéricos del mismo tipo.
  - La eficiencia de la memoria es importante
  - No necesitas operaciones matemáticas complejas
- Utilice matrices NumPy cuando:
  - Necesitas realizar operaciones matemáticas con arrays.
  - Estás trabajando con grandes conjuntos de datos.
  - Necesitas operaciones de difusión y vectorizadas.
  - Necesitas matrices multidimensionales
- Utilice listas de Python cuando:
  - Necesitas una colección de tipos de datos mixtos.
  - Estás agregando/eliminando elementos con frecuencia
  - No necesitas operaciones numéricas de alto rendimiento.
