# Pseudocódigo

El pseudocódigo es la descripción paso a paso del algoritmo escrito en lenguaje natural. Esta diseñado para el entendimiento humano, no para la ejeución de la maquina y no sigue la sintaxis de ningún lenguaje de programación específico.

## ¿Por qué necesitamos pseudocódigo?

- Ayuda a los programadores a planificar y diseñar la solucione antes de escribir el código real.
- Hace que la lógica sea fácil de entender para otros que lean la solución.
- Cierra la brecha entre el diseño de algoritmos y la programación.
- Reduce los errores lógicos antes de la implementación.

## ¿Como escribir pseudocódigo?

Antes de escribir pseudocódigo debes tener en mente los siguientes punto:

- Primero definir el objetivo
- Todo pseudocódigo debe tener un comienzo y un final (INICIO, FIN)
- Organizar la tareas de manera lógica y escribir una secuencia de pasos
- La entrada de datos se representara por LEER
- La salida de datos se representara por MOSTRAR
- Utilice construcciones de programación estándar.
    - IF-ELSE
    - FOR
    - WHILE
- Aplicar la indentación adecuada a los bloques para mejorar la legibilidad y la claridad.
- Utilice nombres claros y significativos para las variables y los pasos.
- Escribir las palabras claves en mayusculas (IF, ELSE, WHILE, FOR)
- Asegurese de que el pseudocódigo es:
    - Completo
    - Finito
    - Facil de entender

### Ejercicio 1

  - **Enunciado:** Crear un algoritmo que permita calcular el área de un rectángulo. El programa debe solicitar al usuario la base y la altura, calcular el área y mostrar el resultado.
  - **Entrada:** base, altura
  - **Proceso:** área = base x altura
  - **Salida:** área

Expresión en pseudocódigo

~~~
INICIO

    LEER base
    LEER altura

    area = base * altura

    MOSTRAR area

FIN
~~~

## Ejercicio 4

Leer un número e indicar si es par o es impar

~~~
INICIO
  LEER numero
  IF numero MOD 2 = 0 THEN
    MOSTRAR "Es Par"
  ELSE
    MOSTRAR "Es Impar"
FIN
~~~

Verificar que un número sea positivo o negativo

~~~
INICIO
  LEER numero
  IF numero >= 0 THEN Por verdad
    MOSTRAR "El número es positivo"
  ELSE
    MOSTRAR "El número es negativo"
FIN
~~~
