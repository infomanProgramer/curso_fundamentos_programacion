# Enfoque modular en la programación

La programación modular consiste en subdividir un programa informático en subprogramas independientes. Un módulo es un componente de software independiente. A menudo se puede utilizar en diversas aplicaciones y funciones junto con otros componentes del sistema.

- Algunos programas pueden tener miles o millones de líneas, y gestionarlos resulta bastante difícil, ya que pueden contener demasiados errores de sintaxis o de lógica. Por ello, para gestionar este tipo de programas se ha recurrido al concepto de programación modular.
- Cada submódulo contiene algo necesario para ejecutar solo un aspecto de la funcionalidad deseada.
- La programación modular hace hincapié en dividir los programas grandes en problemas más pequeños para aumentar la mantenibilidad y la legibilidad del código, y para que el programa sea más fácil de manejar para realizar cambios en el futuro o corregir errores.

### Ventajas de utilizar el enfoque de programación modular

1. **Facilidad de uso:** Este enfoque permite la simplicidad, ya que en lugar de concentrarse en miles o millones de líneas de código de una sola vez, podemos acceder a ellas en forma de módulos. Esto facilita la depuración del código y reduce la probabilidad de errores.
Reutilización: Permite al usuario reutilizar la funcionalidad con una interfaz diferente sin tener que escribir todo el programa.
2. **Reutilización:** Permite al usuario reutilizar la funcionalidad con una interfaz diferente sin tener que volver a escribir todo el programa.
3. **Facilidad de mantenimiento:** Esto ayuda a reducir los conflictos al trabajar en módulos, lo que permite que un equipo colabore adecuadamente al desarrollar una aplicación de gran tamaño.

## Diseño modular eficaz en ingeniería de software

**El papel del diseño modular efectivo en la ingeniería de software:** Cualquier software se compone de muchos sistemas que contienen varios subsistemas, y estos subsistemas a su vez contienen otros subsistemas. Por lo tanto, diseñar un sistema completo de una sola vez, que incluya todas las funcionalidades requeridas, es un trabajo arduo y el proceso puede presentar muchos errores debido a su gran tamaño. Para resolver este problema, el equipo de desarrollo divide el software completo en varios módulos. Un módulo se define como los componentes únicos y direccionables del software que pueden resolverse y modificarse de forma independiente sin perturbar (o afectando mínimamente) a otros módulos del software. Por lo tanto, todo diseño de software debe seguir la modularidad. El proceso de dividir un software en múltiples módulos independientes, donde cada módulo se desarrolla por separado, se denomina modularización. Un diseño modular efectivo se logra si los módulos particionados pueden resolverse, modificarse y compilarse por separado. En este contexto, módulos compilables por separado significa que, después de realizar cambios en un módulo, no es necesario recompilar todo el sistema de software. Para construir un software con un diseño modular efectivo, entra en juego un factor llamado **independencia funcional**. El significado de Independencia Funcional es que una función es atómica, de modo que realiza una única tarea del software con poca o ninguna interacción con otros módulos. La Independencia Funcional se considera un indicador de crecimiento en la modularidad; es decir, una mayor independencia funcional resulta en un sistema de software bien diseñado, y el diseño influye en la calidad del software. Beneficios de los módulos/funciones independientes en el diseño de software: Dado que la funcionalidad del software se ha dividido en niveles atómicos, los desarrolladores obtienen una comprensión clara de los requisitos de cada función, lo que facilita el diseño del software y reduce los errores. Al ser independientes, los módulos tienen una dependencia limitada o casi nula de otros módulos. Por lo tanto, es posible realizar cambios en un módulo sin afectar a todo el sistema. La propagación de errores de un módulo a otro y, posteriormente, a todo el sistema, se puede minimizar, lo que ahorra tiempo durante las pruebas y la depuración. *La independencia de los módulos de un sistema de software se puede medir utilizando dos criterios*:

- Cohesión
- Acoplamiento

Estos se explican a continuación:

La **cohesión** es una medida de la fuerza de la relación entre las distintas funciones dentro de un módulo. Existen 7 tipos que se enumeran a continuación en orden de mayor a menor cohesión:

1. Cohesión funcional
2. Cohesión secuencial
3. Cohesión comunicacional
4. Cohesión procedimental
5. Cohesión temporal
6. Cohesión lógica
7. Cohesión coincidente

El **acoplamiento** es una medida de la fuerza de la relación entre los distintos módulos de un software. Existen 6 tipos que se enumeran a continuación en orden de menor a mayor acoplamiento:

1. Acoplamiento de datos
2. Acoplamiento de sellos
3. Acoplamiento de control
4. Acoplamiento externo
5. Acoplamiento común
6. Acoplamiento de contenido

*Un buen diseño de software requiere una alta cohesión y un bajo acoplamiento.*
