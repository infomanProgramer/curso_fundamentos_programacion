# Programación modular y modulos

La programación modular es una técnica de diseño de software basada en el principio general del diseño modular. Este enfoque ha demostrado ser indispensable en la ingeniería desde mucho antes de la aparición de las primeras computadoras. El diseño modular implica la división de un sistema complejo en partes o componentes más pequeños, es decir, módulos. Estos componentes pueden crearse y probarse de forma independiente y, en muchos casos, incluso pueden utilizarse en otros sistemas.

Hoy en día, prácticamente no existe ningún producto que no dependa en gran medida de la modularización, como los automóviles y los teléfonos móviles. Los ordenadores pertenecen a ese grupo de productos que se modularizan al máximo. Por lo tanto, lo que es esencial para el hardware es una necesidad ineludible para el software que se ejecuta en los ordenadores.

La programación modular es una técnica de diseño de software que divide el código en partes separadas, llamadas módulos. **El objetivo de esta separación es que los módulos tengan pocas o ninguna dependencia entre sí**. En otras palabras: la minimización de dependencias es la meta. Al crear un sistema modular, varios módulos se desarrollan de forma independiente y, en mayor o menor medida, se integran. La aplicación ejecutable se crea mediante su combinación.

## Importación de modulos

En resumen: cualquier archivo con la extensión .py que contenga código Python válido puede considerarse un módulo. No se requiere una sintaxis especial para que un archivo sea un módulo. Un módulo puede contener objetos arbitrarios, como archivos, clases o atributos. Se puede acceder a todos estos objetos después de importarlos. Existen diferentes maneras de importar módulos. Lo demostraremos con el módulo de matemáticas:

~~~python
import math
~~~

El módulo math proporciona constantes y funciones matemáticas, por ejemplo:

- π (math.pi)
- la función seno (math.sin())
- función coseno (math.cos())

A cada atributo o función solo se puede acceder anteponiendo "math." a su nombre:

```python
import math

print(math.pi)
```

```python
#@title { display-mode: "form" }
print(math.sin(math.pi/2))
```

```python
#@title { display-mode: "form" }
print(math.cos(math.pi))
```

Es posible importar más de un módulo en una misma instrucción de importación. En este caso, los nombres de los módulos se separan por comas:

~~~python
import math, random
~~~

Las declaraciones de importación se pueden ubicar en cualquier parte del programa, pero es recomendable colocarlas directamente al principio del mismo.

Si solo se necesitan ciertos objetos de un módulo, podemos importar solo esos:

~~~python
from math import sin, pi
~~~

Los demás objetos, por ejemplo, $cos$, no están disponibles después de esta importación. Podemos acceder a $sin$ y $pi$ directamente, es decir, sin anteponerles $math$. En lugar de importar explícitamente ciertos objetos de un módulo, también es posible importar todo el espacio de nombres del módulo que realiza la importación. Esto se puede lograr utilizando un asterisco en la importación:

~~~python
from math import *

sin(3.01) + tan(cos(2.1)) + e
~~~

No se recomienda usar la notación de asterisco en una instrucción de importación, excepto cuando se trabaja en la consola interactiva de Python. Una razón es que el origen de un nombre puede ser bastante oscuro, ya que no se puede ver de qué módulo se importó. Demostraremos otra complicación importante en el siguiente ejemplo:

```python
from numpy import *
from math import *
print(sin(3))
```

Vamos a modificar ligeramente el ejemplo anterior cambiando el orden de las importaciones:

```python
from math import *
from numpy import *
print(sin(3))
```

```python
sin(3)
```

La gente usa la notación de asterisco porque es muy conveniente. Significa evitar mucho trabajo de escritura tedioso. Otra forma de reducir el esfuerzo de escritura es renombrar un espacio de nombres. Un buen ejemplo de esto es el módulo numpy. Difícilmente encontrarás un ejemplo o un tutorial en el que importen este módulo con la instrucción.

~~~python
import numpy
~~~

Es como una ley no escrita importarlo con

~~~python
import numpy as np
~~~

Ahora puedes anteponer "np." a todos los objetos de numpy en lugar de "numpy.":

```python
import numpy as np
np.diag([3, 11, 7, 9])
```

## Diseñando y escribiendo módulos

Un módulo en Python es simplemente un archivo que contiene definiciones y sentencias de Python. El nombre del módulo se obtiene a partir del nombre del archivo, eliminando la extensión .py. Por ejemplo, si el archivo se llama *fibonacci.*, el módulo se llamará *fibonacci*.

Vamos a convertir operaciones básicas en un módulo:

*operaciones.py*
~~~python
def suma(a, b):
    c = a + b
    return c
def resta(a, b):
    c = a - b
    return c
~~~

*main.py*
~~~python
import operaciones
#from operaciones import suma, resta
#from operaciones import *


print(operaciones.suma(4, 5))
print(operaciones.resta(4, 5))
~~~

## Ejecución de módulos como scripts

En esencia, un módulo de Python es un script, por lo que puede ejecutarse como tal:

~~~
python fibo.py
~~~

El módulo que se ha iniciado como script se ejecutará como si se hubiera importado, pero con una excepción: la variable de sistema `name` se establece en "main". Por lo tanto, es posible programar un comportamiento diferente en un módulo para ambos casos. Con la siguiente instrucción condicional, el archivo se puede usar como módulo o como script, pero solo si se ejecuta como script se iniciará el método `fib` con un argumento de línea de comandos:

~~~python
if __name__ == "__main__":
  import sys
  fib(int(sys.argv[1]))
~~~

## Renombrando un Namespace

Al importar un módulo, se puede cambiar el nombre del espacio de nombres:

~~~python
import math as mathematics
print(mathematics.cos(mathematics.pi))
~~~

Después de esta importación existe un espacio de nombres mathematics pero no un espacio de nombres math. Es posible importar solo algunos métodos de un módulo:

```python
import math as mathematics
print(mathematics.cos(mathematics.pi))
print(math.pi)
```

# Paquetes

## Introducción

Aprendimos que los módulos son archivos que contienen instrucciones y definiciones de Python, como definiciones de funciones y clases. En este capítulo, aprenderemos cómo agrupar varios módulos para formar un paquete.

Un paquete es básicamente un directorio con archivos Python y un archivo llamado __init__.py. Esto significa que cualquier directorio dentro de la ruta de Python que contenga un archivo llamado __init__.py será tratado como un paquete por Python. Es posible incluir varios módulos en un paquete.

Los paquetes son una forma de estructurar el espacio de nombres de módulos de Python mediante nombres de módulos separados por puntos. A.B representa un submódulo llamado B en un paquete llamado A. Dos paquetes diferentes, como P1 y P2, pueden tener módulos con el mismo nombre, por ejemplo, A. El submódulo A del paquete P1 y el submódulo A del paquete P2 pueden ser completamente diferentes. Un paquete se importa como un módulo "normal". Comenzaremos este capítulo con un ejemplo sencillo.
