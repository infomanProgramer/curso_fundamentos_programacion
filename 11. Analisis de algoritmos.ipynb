# **11. Análisis de algoritmos**

El análisis de algoritmos es un aspecto fundamental de la informática que implica evaluar el rendimiento de algoritmos y programas. La eficiencia se mide en términos de tiempo y espacio.

### **¿Por qué es importante el análisis de algoritmos?**

Hay muchos aspectos importantes que deben tenerse en cuenta, como la facilidad de uso, la modularidad, la seguridad, la mantenibilidad, etc. ¿Por qué preocuparse por el rendimiento?

- La respuesta es sencilla: solo podemos tener todo lo anterior si tenemos rendimiento. Por lo tanto, el rendimiento es como la moneda con la que podemos adquirir todo lo mencionado.
- Otra razón para estudiar el rendimiento es que *¡la velocidad es divertida!*

### ¿Por qué es importante el análisis de algoritmos?

El análisis de algoritmos es una parte importante de la teoría de la complejidad computacional, que proporciona una estimación teórica de los recursos necesarios para que un algoritmo resuelva un problema computacional específico. El análisis de algoritmos consiste en determinar la cantidad de tiempo y recursos espaciales necesarios para su ejecución.

- Para predecir el comportamiento de un algoritmo con grandes conjuntos de datos de entrada (Software Escalable).
- Resulta mucho más conveniente disponer de medidas sencillas para evaluar la eficiencia de un algoritmo que implementarlo y probar su eficiencia cada vez que cambia algún parámetro del sistema informático subyacente.
- Y lo que es más importante, al analizar diferentes algoritmos, podemos compararlos para determinar cuál es el mejor para nuestro propósito.

### **Orden de crecimiento**

Sean $f(n)$ y $g(n)$ los tiempos que tardan dos algoritmos, donde:

$n >= 0$ y $f(n)$ y $g(n)$ también son mayores o iguales que $0$.

Se dice que una función $f(n)$ crece más rápido que $g(n)$ si $g(n)/f(n)$ cuando $n$ tiende a infinito es $0$ (o $f(n)/g(n)$ cuando $n$ tiende a infinito es infinito).

**Ejemplo 1**:
~~~
f(n) = 1000,  g(n) = n + 1
Para n > 999, g(n) siempre será más grande que f(n) por que el orden de crecimiento de  g(n) es más grande que f(n).
~~~

**Ejemplo 2**
~~~
f(n) = 4n2 ,  g(n) = 2n + 2000
f(n) tiene un orden de crecimiento superior ya que crece cuadráticamente en términos del tamaño de entrada.
~~~
