# **tkinter - Interfaz de Python para Tcl/Tk**

El paquete tkinter (“interfaz Tk”) es la interfaz estándar de Python para el conjunto de herramientas GUI Tcl/Tk. Tanto Tk como tkinter están disponibles en la mayoría de las plataformas Unix, incluyendo macOS, así como en sistemas Windows.

Al ejecutar 

~~~
python -m tkinter 
~~~

Se abrirá una ventana que muestra una interfaz Tk sencilla, lo que le permitirá saber que tkinter está correctamente instalado en su sistema y también le mostrará qué versión de Tcl/Tk está instalada, para que pueda leer la documentación de Tcl/Tk específica para esa versión.

## **Arquitectura**

Tcl/Tk no es una única biblioteca, sino que consta de varios módulos distintos, cada uno con funcionalidad propia y su documentación oficial. Las versiones binarias de Python también incluyen un módulo adicional.

- **Tcl** es un ***lenguaje de programación*** interpretado dinámico, al igual que Python. Si bien puede utilizarse de forma independiente como lenguaje de programación de propósito general, lo más común es integrarlo en aplicaciones C como motor de scripting o interfaz para el conjunto de herramientas Tk. La biblioteca Tcl cuenta con una interfaz C para crear y gestionar una o más instancias de un intérprete Tcl, ejecutar comandos y scripts Tcl en dichas instancias y añadir comandos personalizados implementados en Tcl o C. Cada intérprete dispone de una cola de eventos, y existen mecanismos para enviar y procesar eventos. A diferencia de Python, el modelo de ejecución de Tcl está diseñado en torno a la multitarea cooperativa, y Tkinter salva esta diferencia (véase el modelo de subprocesos para más detalles).
- **Tk** es un paquete de Tcl implementado en C que añade ***comandos personalizados para crear y manipular widgets*** de interfaz gráfica de usuario (GUI). Cada objeto Tk incorpora su propia instancia del intérprete de Tcl con Tk cargado. Los widgets de Tk son altamente personalizables, aunque a costa de una apariencia algo anticuada. Tk utiliza la cola de eventos de Tcl para generar y procesar eventos de la GUI.
- **Themed Tk (Ttk)** es una **familia más reciente de widgets** de Tk que ofrece una apariencia mucho mejor en diferentes plataformas que muchos de los widgets clásicos de Tk. Ttk se distribuye como parte de Tk, a partir de la versión 8.5. Las interfaces para Python se proporcionan en un módulo independiente, tkinter.ttk.

Internamente, Tk y Ttk utilizan las funcionalidades del sistema operativo subyacente, es decir, Xlib en Unix/X11, Cocoa en macOS y GDI en Windows.

Cuando una aplicación Python utiliza una clase de Tkinter, por ejemplo, para crear un widget, el módulo tkinter primero ensambla una cadena de comandos Tcl/Tk. Esta cadena se pasa a un módulo binario interno _tkinter, que a su vez llama al intérprete Tcl para evaluarla. El intérprete Tcl llama entonces a los paquetes Tk y/o Ttk, que a su vez llaman a Xlib, Cocoa o GDI.

## **Modulos Tkinter**

La compatibilidad con Tkinter se distribuye en varios módulos. La mayoría de las aplicaciones necesitarán el módulo principal de tkinter, así como el módulo tkinter.ttk, que proporciona el conjunto de widgets con temas modernos y la API:

Los módulos que brindan soporte para Tk incluyen:

- **tkinter:** Módulo principal de Tkinter.
- **tkinter.colorchooser:** Cuadro de diálogo para que el usuario elija un color.
- **tkinter.filedialog:** Cuadros de diálogo comunes que permiten al usuario especificar un archivo para abrir o guardar.
- **tkinter.font:** Utilidades para facilitar el trabajo con fuentes tipográficas.
- **tkinter.scrolledtext:** Widget de texto con barra de desplazamiento vertical integrada.
- **tkinter.simpledialog:** Diálogos básicos y funciones prácticas.
- **tkinter.ttk:** Conjunto de widgets temáticos introducido en Tk 8.5, que proporciona alternativas modernas para muchos de los widgets clásicos del módulo principal de tkinter.

Módulos adicionales:

- **_tkinter:** Módulo binario que contiene la interfaz de bajo nivel para Tcl/Tk. Se importa automáticamente mediante el módulo principal de tkinter y los programadores de aplicaciones nunca deben usarlo directamente. Generalmente es una biblioteca compartida (o DLL), pero en algunos casos puede estar enlazada estáticamente con el intérprete de Python.
- **idlelib:** Entorno Integrado de Desarrollo y Aprendizaje (IDLE) de Python. Basado en tkinter.
- **tkinter.constants:** Constantes simbólicas que pueden usarse en lugar de cadenas de texto al pasar diversos parámetros a las llamadas de Tkinter. Se importan automáticamente mediante el módulo principal de Tkinter.
- **tkinter.dnd:** (Experimental) Compatibilidad con arrastrar y soltar para tkinter. Esta función quedará obsoleta cuando se reemplace por la función DND de Tk.
- **turtle:** Gráficos de tortuga en una ventana Tk.


## **Conservador de vida Tkinter**

#### Un programa Hola Mundo

Comenzaremos con un ejemplo sencillo de aplicación "Hola Mundo" en Tkinter. No es la aplicación más simple que podríamos escribir, pero tiene lo suficiente para ilustrar algunos conceptos clave que necesitarás conocer.

~~~python
from tkinter import *
from tkinter import ttk
root = Tk()
frm = ttk.Frame(root, padding=10)
frm.grid()
ttk.Label(frm, text="Hello World!").grid(column=0, row=0)
ttk.Button(frm, text="Quit", command=root.destroy).grid(column=1, row=0)
root.mainloop()
~~~

- Tras las importaciones, la siguiente línea crea una instancia de la clase Tk, que inicializa Tk y crea su intérprete Tcl asociado. También crea una ventana de nivel superior, conocida como ventana raíz, que sirve como ventana principal de la aplicación.
- La siguiente línea crea un widget de marco, que en este caso contendrá una etiqueta y un botón que crearemos a continuación. El marco se ajusta dentro de la ventana raíz.
- A continuación, se crea un widget de botón y se coloca a la derecha de la etiqueta. Al pulsarlo, se llamará al método destroy() de la ventana raíz.
- Finalmente, el método mainloop() muestra todo en la pantalla y responde a la entrada del usuario hasta que finaliza el programa.

#### Conceptos importantes de Tk

Incluso este sencillo programa ilustra los siguientes conceptos clave de Tk:

- **widgets** Una interfaz de usuario de Tkinter se compone de widgets individuales. Cada widget se representa como un objeto Python, instanciado a partir de clases como ttk.Frame, ttk.Label y ttk.Button.
- **jerarquía de widgets** Los widgets se organizan jerárquicamente. La etiqueta y el botón se encuentran dentro de un marco, que a su vez está dentro de la ventana raíz. Al crear cada widget hijo, su widget padre se pasa como primer argumento al constructor del widget.
- **opciones de configuración** Los widgets tienen opciones de configuración que modifican su apariencia y comportamiento, como el texto que se muestra en una etiqueta o botón. Las diferentes clases de widgets tendrán diferentes conjuntos de opciones.
- **gestión de la geometría** Los widgets no se agregan automáticamente a la interfaz de usuario al crearse. Un gestor de geometría como Grid controla su ubicación en la interfaz de usuario.
- **bucle de eventos** Tkinter reacciona a la entrada del usuario, a los cambios del programa e incluso actualiza la pantalla solo cuando se está ejecutando un bucle de eventos. Si el programa no está ejecutando el bucle de eventos, la interfaz de usuario no se actualizará.

#### Comprender cómo Tkinter envuelve Tcl/Tk

~~~python
ttk::frame .frm -padding 10
grid .frm
grid [ttk::label .frm.lbl -text "Hello World!"] -column 0 -row 0
grid [ttk::button .frm.btn -text "Quit" -command "destroy ."] -column 1 -row 0
~~~

La sintaxis de Tcl es similar a la de muchos lenguajes de shell, donde la primera palabra es el comando que se va a ejecutar, seguido de los argumentos de dicho comando, separados por espacios. Sin entrar en demasiados detalles, observe lo siguiente:

- Los comandos utilizados para crear widgets **(como ttk::frame)** corresponden a las clases de widgets en Tkinter.
- Las opciones de los widgets de Tcl **(como -text)** se corresponden con los argumentos de palabra clave en Tkinter.
- En Tcl, los widgets se referencian mediante una ruta de archivo **(como .frm.btn)**, mientras que Tkinter no utiliza nombres, sino referencias a objetos.
- La posición de un widget en la jerarquía de widgets está codificada en su ruta (jerárquica), que utiliza un punto (.) como separador de ruta. La ruta para la ventana raíz es simplemente un punto (.). En Tkinter, la jerarquía se define no por la ruta, sino especificando el widget padre al crear cada widget hijo.
- Las operaciones que se implementan como comandos independientes en Tcl (como grid o destroy) se representan como métodos en los objetos widget de Tkinter. Como verá en breve, en otras ocasiones Tcl utiliza lo que parecen ser llamadas a métodos en los objetos widget, que se asemejan más a lo que se usa en Tkinter.

#### ¿Cómo hago...? ¿Qué opción...?

Si no estás seguro de cómo hacer algo en Tkinter y no lo encuentras de inmediato en el tutorial o la documentación de referencia que estás utilizando, existen algunas estrategias que pueden resultarte útiles.

En primer lugar, ten en cuenta que el funcionamiento de los widgets individuales puede variar según la versión de Tkinter y Tcl/Tk. Si buscas documentación, asegúrate de que corresponda a las versiones de Python y Tcl/Tk instaladas en tu sistema.

Al buscar cómo usar una API, es útil conocer el nombre exacto de la clase, opción o método que se está utilizando. La introspección, ya sea en una consola interactiva de Python o con print(), puede ayudar a identificar lo que se necesita.

Para saber qué opciones de configuración están disponibles en cualquier widget, llame a su método **configure()**, que devuelve un diccionario con diversa información sobre cada objeto, incluidos sus valores predeterminados y actuales. Use **keys()** para obtener solo los nombres de cada opción.

~~~python
btn = ttk.Button(frm, ...)
print(btn.configure().keys())
~~~

Como la mayoría de los widgets comparten muchas opciones de configuración, puede resultar útil averiguar cuáles son específicas de una clase de widget en particular. Una forma de hacerlo es comparar la lista de opciones con la de un widget más sencillo, como un marco.

~~~python
print(set(btn.configure().keys()) - set(frm.configure().keys()))
~~~

De forma similar, puedes encontrar los métodos disponibles para un objeto widget usando la función estándar dir(). Si lo intentas, verás que hay más de 200 métodos comunes para widgets, por lo que identificar los específicos de una clase de widget resulta útil.

~~~python
print(dir(btn))
print(set(dir(btn)) - set(dir(frm)))
~~~

Cómo navegar por el manual de referencia de Tcl/Tk

Como se mencionó, el manual de referencia oficial de comandos de Tk (páginas man) suele ser la descripción más precisa de lo que hacen las operaciones específicas en los widgets. Incluso cuando se conoce el nombre de la opción o el método necesario, es posible que aún haya que consultar otros lugares.

Si bien todas las operaciones en Tkinter se implementan como llamadas a métodos en objetos widget, habrás visto que muchas operaciones de Tcl/Tk aparecen como comandos que toman una ruta de acceso a un widget como primer parámetro, seguida de parámetros opcionales, por ejemplo

~~~python
destroy .
grid .frm.btn -column 0 -row 0
~~~

Otros, sin embargo, se parecen más a métodos llamados en un objeto widget (de hecho, cuando creas un widget en Tcl/Tk, crea un comando Tcl con el nombre de la ruta del widget, siendo el primer parámetro de ese comando el nombre de un método a llamar).

~~~python
.frm.btn invoke
.frm.lbl configure -text "Goodbye"
~~~

En la documentación de referencia oficial de Tcl/Tk, encontrará la mayoría de las operaciones que parecen llamadas a métodos en la página man de un widget específico (por ejemplo, encontrará el método invoke() en la página man de ttk::button), mientras que las funciones que toman un widget como parámetro a menudo tienen su propia página man (por ejemplo, grid).

También descubrirás que muchos métodos de Tkinter tienen nombres compuestos, por ejemplo, winfo_x(), winfo_height(), winfo_viewable(). Encontrarás la documentación de todos ellos en la página man de winfo.

#### Modelo de enhebrado

Python y Tcl/Tk tienen modelos de subprocesos muy diferentes, que tkinter intenta solucionar. Si utilizas subprocesos, es posible que debas tener esto en cuenta.

Cada objeto Tk creado por tkinter contiene un intérprete Tcl. Además, registra qué hilo creó dicho intérprete. Se pueden realizar llamadas a tkinter desde cualquier hilo de Python. Internamente, si una llamada proviene de un hilo distinto al que creó el objeto Tk, se publica un evento en la cola de eventos del intérprete y, al ejecutarse, el resultado se devuelve al hilo de Python que realizó la llamada.

Las aplicaciones Tcl/Tk suelen estar orientadas a eventos, lo que significa que, tras la inicialización, el intérprete ejecuta un bucle de eventos (Tk.mainloop) y responde a ellos. Dado que se ejecuta en un solo hilo, los manejadores de eventos deben responder con rapidez; de lo contrario, bloquearán el procesamiento de otros eventos. Para evitarlo, los cálculos de larga duración no deben ejecutarse en un manejador de eventos, sino que deben dividirse en partes más pequeñas mediante temporizadores o ejecutarse en otro hilo. Esto difiere de muchos kits de herramientas de interfaz gráfica de usuario (GUI), donde la GUI se ejecuta en un hilo completamente independiente del resto del código de la aplicación, incluidos los manejadores de eventos.

Si el intérprete de Tcl no está ejecutando el bucle de eventos ni procesando eventos, cualquier llamada a tkinter realizada desde subprocesos distintos al que ejecuta el intérprete de Tcl fallará.

## **Crea tu primera aplicación GUI con Tkinter.**


### Clase Tk()

La **clase Tk()** se utiliza para crear la ventana principal de la aplicación

~~~python
root = Tk(screenName=None, baseName=None, className='Tk', useTk=1)
~~~

**Parametros**

- **screenName (opcional):** Especifica la pantalla (se usa principalmente en sistemas Unix con varias pantallas).
- **baseName (opcional):** Establece el nombre base para la aplicación (el valor predeterminado es el nombre del script).
- **className (opcional):** Define el nombre de la clase de ventana (utilizada para el estilo y la configuración del administrador de ventanas).
- **className (opcional):** Define el nombre de la clase de ventana (utilizada para el estilo y la configuración del administrador de ventanas).

### Mainloop()

El método **mainloop()** inicia el bucle de eventos de una aplicación Tkinter. Mantiene la ventana abierta, espera acciones del usuario (como clics del ratón o pulsaciones de teclas) y procesa los eventos hasta que se cierra la ventana. A continuación se muestra la sintaxis:

~~~python
root.mainloop()
~~~

**Ejemplo**

~~~python
import tkinter as tk

root = tk.Tk()
# Widgets are added here

root.mainloop()
~~~

### Widget de Tkinter

Existen varios widgets de Tkinter que podemos incorporar a nuestra aplicación. A continuación, se explican algunos de los widgets más importantes:

#### **Label**

El widget Label se utiliza para mostrar texto o imágenes en una ventana de Tkinter. Se usa comúnmente para mostrar mensajes, títulos o instrucciones. A continuación se muestra la sintaxis:

~~~python
label = tk.Label(master, option=value)
~~~

**Parámetros** 

- **master:** se refiere a la ventana o contenedor principal en el que se coloca la etiqueta.

**Ejemplo**

~~~python
import tkinter as tk

root = tk.Tk()
label = tk.Label(root, text="Curso de programación")
label.pack()

root.mainloop()
~~~

**Explicación:** label.pack() coloca la etiqueta dentro de la ventana.

#### **Button**

El widget Botón es un componente en el que se puede hacer clic para realizar una acción al pulsarlo, como enviar datos o cerrar una ventana. A continuación se muestra la sintaxis:

~~~python
button = tk.Button(master, option=value)
~~~

**Ejemplo:** Este ejemplo crea una ventana titulada “Contando segundos” con un botón de Detener. Al hacer clic en el botón, la aplicación se cierra.

~~~python
import tkinter as tk

root = tk.Tk()
root.title("Counting Seconds")

button = tk.Button(root, text="Stop", width=50, command=root.destroy)
button.pack()

root.mainloop()
~~~

**Explicación:**

- **tk.Button():** Crea un widget de botón.
- **comando=root.destroy:** Cierra la aplicación al hacer clic en el botón.
- **botón.pack():** Coloca el botón en la ventana.

#### **Entry o campos de texto**

El widget Entrada se utiliza para aceptar texto de una sola línea del usuario. Para texto de varias líneas, se utiliza el widget Texto. A continuación se muestra la sintaxis:

~~~python
entry = tk.Entry(master, option=value)
~~~

**Explicación:**

- **Master:** La ventana o marco principal en el que se mostrará el widget.
- **Option:** Las distintas opciones que ofrece el widget de entrada son:
    - **bg :** El color de fondo normal que se muestra detrás de la etiqueta y el indicador.
    - **bd :** El tamaño del borde alrededor del indicador. El valor predeterminado es de 2 píxeles.
    - **font :** La fuente utilizada para el texto.
    - **fg :** El color utilizado para representar el texto.
    - **justify :** Si el texto contiene varias líneas, esta opción controla cómo se justifica el texto: CENTER, LEFT, or RIGHT.
    - **relief :** Con el valor predeterminado, relief=FLAT. Puede configurar esta opción a cualquiera de los otros estilos, como: SUNKEN, RIGID, RAISED, GROOVE 
    - **show :** Normalmente, los caracteres que el usuario escribe aparecen en la entrada. Para crear una entrada .password. que muestre cada carácter como un asterisco, configure show="*". 
    - **textvariable :** Para poder recuperar el texto actual de su widget de entrada, debe establecer esta opción en una instancia de la clase StringVar.
- **Métodos**

**Ejemplo**: Los distintos métodos que ofrece el widget de entrada son:

- **get():** Devuelve el texto actual de la entrada como una cadena de texto.
- **delete():** Elimina caracteres del widget.
- **insert (index, 'name') :** Inserta la cadena 'nombre' antes del carácter en el índice dado.

Este ejemplo crea un formulario sencillo con campos de entrada para Nombre y Apellido. El gestor de geometría grid() se utiliza para organizar las etiquetas y los cuadros de entrada en un diseño tipo tabla.

~~~python
import tkinter as tk

root = tk.Tk()

tk.Label(root, text="Nombres").grid(row=0, column=0)
tk.Label(root, text="Apellidos").grid(row=1, column=0)

entry1 = tk.Entry(root)
entry2 = tk.Entry(root)

entry1.grid(row=0, column=1)
entry2.grid(row=1, column=1)

root.mainloop()
~~~

**Explicación:**

- **tk.Entry():** Crea un cuadro de entrada de texto de una sola línea.
- **grid():** Organiza los widgets en filas y columnas.

#### **CheckButton**

El widget Checkbutton se utiliza para crear una casilla de verificación que se puede activar o desactivar. Generalmente se vincula a una variable para almacenar su estado actual. A continuación se muestra la sintaxis:

~~~python
check = tk.Checkbutton(master, option=value)
~~~

**Ejemplo**

Este ejemplo crea una ventana con dos casillas de verificación etiquetadas como Masculino y Femenino. Cada casilla está conectada a una variable IntVar() que almacena su estado (1 para marcada, 0 para desmarcada).

~~~python
import tkinter as tk

root = tk.Tk()

var1 = tk.IntVar()
var2 = tk.IntVar()

tk.Checkbutton(root, text="Male", variable=var1).grid(row=0, sticky=tk.W)
tk.Checkbutton(root, text="Female", variable=var2).grid(row=1, sticky=tk.W)

root.mainloop()
~~~

**Explicación:**

- **tk.IntVar():** Almacena el estado de la casilla de verificación (1 o 0).
- **grid():** Alinea los widgets en filas y columnas.
- **sticky=tk.W:** Alinea la casilla de verificación a la izquierda.

#### **Radiobutton**

El widget RadioButton permite al usuario seleccionar una única opción de un grupo de opciones. Todos los botones de radio de un grupo comparten la misma variable. A continuación se muestra la sintaxis:

~~~python
radio = tk.Radiobutton(master, option=value)
~~~

**Ejemplo**

Este ejemplo crea dos botones de opción etiquetados como A y B. Solo se puede seleccionar una opción a la vez.

~~~python
import tkinter as tk

root = tk.Tk()
v = tk.IntVar()

tk.Radiobutton(root, text="A", variable=v, value=1).pack(anchor=tk.W)
tk.Radiobutton(root, text="B", variable=v, value=2).pack(anchor=tk.W)
root.mainloop()
~~~

**Explicación:**

- **tk.Radiobutton():** Creates a radio button widget.
- **tk.IntVar():** Stores the selected option value.
- **value:** Assigned value when the radio button is selected.
- **anchor=tk.W:** Aligns buttons to the left.

#### **Listbox**

El widget Listbox muestra una lista de elementos de la cual el usuario puede seleccionar una o más opciones. A continuación se muestra la sintaxis:

~~~python
listbox = tk.Listbox(master, option=value)
~~~

**Ejemplo**

El widget Listbox muestra una lista de elementos de la cual el usuario puede seleccionar una o más opciones. A continuación se muestra la sintaxis:

~~~python
import tkinter as tk

root = tk.Tk()

lb = tk.Listbox(root)
lb.insert(1, "Python")
lb.insert(2, "Java")
lb.insert(3, "C++")
lb.insert(4, "Any other")

lb.pack()
root.mainloop()
~~~
