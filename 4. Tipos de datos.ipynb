# Tipos de datos en programación

En programación, el tipo de dato es un atributo asociado a un dato que indica al sistema informático cómo interpretar su valor. Comprender los tipos de datos garantiza que estos se recopilen en el formato preferido y que el valor de cada propiedad sea el esperado.



image.png

## ¿Qué son los tipos de datos en programación?

*"Un atributo que identifica un dato e indica a un sistema informático cómo interpretar su valor se denomina tipo de dato"*

En programación, el término "tipo de dato" describe el valor que posee una variable y las operaciones matemáticas, relacionales o lógicas que se pueden realizar sobre ella sin generar errores. Numerosos lenguajes de programación, por ejemplo, utilizan los tipos de datos como:

- cadena
- entero
- coma flotante,
- números enteros
- valores con decimales

Un intérprete o compilador puede determinar cómo un programador planea usar un conjunto de datos consultando su tipo de dato.

Los datos se presentan en diferentes formatos. Algunos ejemplos son:

- Tu nombre: una cadena de caracteres
- Tu edad: generalmente un número entero
- La cantidad de dinero que llevas en el bolsillo: generalmente decimal
- La fecha de hoy: escrita en formato de fecha y hora

## Tipo des datos comunes en programación

1. **Tipos de datos primitivos:** Los tipos primitivos son tipos de datos predefinidos, independientes de otros tipos, que incluyen valores básicos de atributos específicos, como texto o valores numéricos. Son el tipo más fundamental y sirven de base para tipos de datos más complejos. La mayoría de los lenguajes de programación probablemente emplean alguna variante de estos tipos de datos simples.

|Tipo de dato|Definición|Ejemplos|
|------------|----------|--------|
|Integer (int)|representa un tipo de datos numéricos para números sin fracciones|300, -200|
|Punto flotante (float)|representa un tipo de datos numéricos para números con fracciones|3.67, 55.88, -78.02|
|Caracteres (char)|representa una sola letra, dígito, signo de puntuación, símbolo o espacio en blanco|a, 5, !|
|Boolean (bool)|Son los valores TRUE o FALSE|TRUE (1), FALSE(0)|
|Date|Fecha en formato YYYY-MM-DD (Según la sintaxis ISO 8601)|2024-01-01|
|Time|Tiempo en el formato **hh:mm:ss** para la hora del día|12:40:00|
|Datetime|Fecha y hora juntos en formato YYYY-MM-DD hh:mm:ss|2024 -01-01 12:34:20|

2. **Tipos de datos compuestos:** Los tipos de datos compuestos se forman a partir de diversos tipos primitivos que normalmente proporciona el usuario. También se les conoce como tipos de datos definidos por el usuario o no primitivos. Los tipos compuestos se dividen en cuatro categorías principales: semiestructurados (almacenan los datos como un conjunto de relaciones); multimedia (almacenan los datos como imágenes, música o vídeos); homogéneos (requieren que todos los valores sean del mismo tipo de dato); y tabulares (almacenan los datos en formato tabular).

|Tipos de datos|Definición|Ejemplo|
|--------------|----------|-------|
|String (string)|Secuencia de caracteres, digitos o simbolos, siempre tratados como texto|"Hola", "18Mb de ram"|
|array (vectores)|Lista de números de elementos ordenados de forma especifica, tipicamente son del mismo tipo|arr[4] = [0, 1, 2, 3]|
|pointers|Los bloques de memoria que se asignan, se gestionan y almacenan dinámicamente|*ptr = 9|

3. **Tipos de datos definidos por el usuario:** Un tipo de dato definido por el usuario (UDT, por sus siglas en inglés) es un tipo de dato derivado de un tipo de dato existente. Puede utilizar otros tipos integrados ya disponibles y crear sus propios tipos de datos personalizados.

|Tipos de datos|Definición|Ejemplo|
|--------------|----------|-------|
|Enumerated Type (enum)|Sirve para representar una variable que solo puede tomar un conjunto de datos definidos previamente|Dia: Domingo, Lunes, Martes, Miercoles, Jueves, Viernes, Sabado|
|Structure|permite la combinación de diferentes tipos de elementos|struct s{...}|
|Union|Contiene un grupo de objetos de datos que pueden tener distintos tipos de datos.|union u {...}|
