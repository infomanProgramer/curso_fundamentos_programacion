# Diagramas de Flujo

Los diagramas de flujo son una representación visual de un algoritmo o un proceso. Los diagramas de flujo usan simbolos, formas como flechas, rectangulos, etc. Los diagramas de flujo tienen sus casos de uso en varios campos como el desarrollo de software, modelado del proceso de negocios e ingenieria.


## Tipos de diagramas de flujo

Existen muchos tipos de diagramas de flujo, cada uno diseñado para representar diferentes tipos de procesos e información. Algunos tipos comunes de diagramas de flujo son:

- **Diagrama de flujo de procesos:** Representa la secuencia de pasos en un proceso. Se utilizan con frecuencia en el modelado de procesos de negocio, la fabricación y la gestión de proyectos.
- **Diagrama de flujo de decisiones:** Se centra en representar los puntos de decisión dentro de un proceso y los posibles resultados de cada decisión. Se utiliza en escenarios de toma de decisiones.

5. **Conector/Referencia:** Cuando un diagrama de flujo se vuelve complejo, es útil usar conectores para evitar confusiones. Los conectores se utilizan para indicar una transición de una parte del diagrama a otra sin necesidad de dibujar líneas largas o complicadas. Las líneas de flujo indican la secuencia exacta en la que se ejecutan las instrucciones. Las flechas representan la dirección del flujo de control y la relación entre los diferentes símbolos del diagrama de flujo.

## Reglas para la creación de diagramas de flujo

Un diagrama de flujo es una representación gráfica de un algoritmo. Debe seguir ciertas reglas al crear un diagrama de flujo.

- La instrucción de apertura del diagrama de flujo debe ser la palabra clave *INICIO*.
- La instrucción final del diagrama de flujo debe ser la palabra clave *FIN*.
- Todos los símbolos del diagrama de flujo deben estar conectados mediante una flecha.
- Cada punto de decisión debe tener dos o más resultados distintos.
- El flujo generalmente debe moverse de arriba hacia abajo o de izquierda a derecha.

## Multicondicional

Un decisión multicondicional permite que un programa evalúe varias condiciones y ejecute una acción diferente dependiendo de cual condición se cumpla

*Si tenemos varias posibilidades, el programa va revisando las condiciones hasta encontrar una que se cumpla*

## Ejemplo 1

~~~
INICIO
  Leer nota
  IF nota >= 90 THEN
    MOSTRAR "Excelente"
  IF nota >= 70 THEN
    MOSTRAR "Muy bien"
  IF nota >= 51 THEN
    MOSTRAR "Aprobado"
  ELSE
    MOSTRAR "Reprobado"
FIN
~~~
